```python
import numpy as np
import plotly.graph_objects as go
from IPython.display import display, clear_output
import time
import imageio
import io

# Lista para acumular fotogramas
frames = []

# Inicializar datos
x_data = []
y_data = []

start = time.time()

for _ in range(100):
    t = time.time() - start
    temp = np.sin(t) + np.random.normal(0, 0.1)

    x_data.append(t)
    y_data.append(temp)

    if len(x_data) > 100:
        x_data.pop(0)
        y_data.pop(0)

    # Crear gráfico nuevo en cada iteración
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=x_data, y=y_data, mode='lines', name='Temperatura'))
    fig.update_layout(
        title='Simulación de Temperatura en Tiempo Real',
        xaxis_title='Tiempo (s)',
        yaxis_title='Temperatura',
        yaxis=dict(range=[-1.5, 1.5])
    )

    # Mostrar en vivo
    clear_output(wait=True)
    display(fig)

    # Capturar fotograma en memoria (usando Kaleido)
    img_bytes = fig.to_image(format="png", width=600, height=400)
    frames.append(imageio.v3.imread(img_bytes))

    time.sleep(0.1)

# Al terminar el bucle, guardo el GIF
output_path = "simulacion_temperatura.gif"
imageio.mimsave(output_path, frames, fps=40)
print(f"GIF guardado en: {output_path}")
```
