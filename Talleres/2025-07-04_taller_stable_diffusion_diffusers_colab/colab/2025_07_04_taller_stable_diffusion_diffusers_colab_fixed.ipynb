```python
from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")
```

```python
generator=torch.Generator(device="cuda").manual_seed(1234)
```

```python
prompt = "A surreal futuristic city in the clouds, digital art"
image = pipe(prompt, num_inference_steps=50, guidance_scale=7.5).images[0]
image.save("output.png")
```

```python
prompt = "A Cyberpunk cow"
image = pipe(prompt, num_inference_steps=50, guidance_scale=7.5).images[0]
image.save("output1.png")
```

```python
prompt = "A photograph of an astronaut riding a horse"
out = pipe(
    prompt,
    negative_prompt="low quality, text, watermark",
    num_inference_steps=30,
    guidance_scale=6.5,
    generator=generator
)
```

```python
import torchvision.transforms as T
to_tensor = T.ToTensor()
tensor_imgs = [to_tensor(img) for img in out.images]
```

```python
import torchvision
grid = torchvision.utils.make_grid(tensor_imgs, nrow=2)
torchvision.utils.save_image(grid, "gallery.png")
```
